```python
# Check missing values first
print(df.isna().sum())

# Impute Age with mean/median
rounded_age = round(df['Age'].mean(), 0)
print("Rounded Age:", rounded_age)

df['Age'] = df['Age'].fillna(rounded_age)
print("\nMissing values after Age imputation:")
print(df.isna().sum())

# Impute Embarked with mode (most common value)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
print("\nMissing values after Embarked imputation:")
print(df.isna().sum())

# Display dataset
df
```

```python
filtered_df = df[(df['Age'] > 40) & (df['Fare'] < 50)]

plt.scatter(filtered_df['Age'], filtered_df['Fare'], color='g')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age v/s Fare (Age>40 & Fare<50)")
plt.show()
```

```python
# Data Encoding
print("Initial row:")
print(df.head(1))

# Replace Sex values cleaner using a dictionary
df['Sex'] = df['Sex'].replace({'male': 'M', 'female': 'F'})

print("\nUpdated Head:")
print(df.head())

print("\nSex Value Counts:")
print(df['Sex'].value_counts())

print("\nSummary Statistics:")
print(df.describe())

print("\nDataFrame Info:")
df.info()

# Drop 'Cabin' column safely if it still exists (too many missing values)
df = df.drop('Cabin', axis=1, errors='ignore')

# Display updated dataframe
df
```

```python
df['Sex'].unique()
df['Embarked'].unique()
df['Pclass'].unique()

from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()

df_array = ohe.fit_transform(df[['Sex', 'Embarked', 'Pclass']]).toarray()
df_array

ohe.categories_

single_list = [cat for main_array in ohe.categories_ for cat in main_array]
single_list

df_new = pd.DataFrame(df_array, columns=single_list)
df_new

df_ml = pd.concat([df, df_new], axis=1)
df_ml
```

```python
# StandardScaler
data = df[['Age', 'Fare']]
data

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().fit(data)
print(scaler)

data_scaled = scaler.transform(data)
print(data_scaled)
print(data_scaled.mean(axis=0))
print(data_scaled.std(axis=0))

scaled_data_set = pd.DataFrame(data_scaled, columns=data.columns)
scaled_data_set
```

```python
# MinMaxScaler
data = df[['Age', 'Fare']]
data

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler().fit(data)
print(scaler)
scaler.data_min_
scaler.data_max_
data.describe()
scaler.feature_range

data_scaled_mm = scaler.transform(data)
print(data_scaled_mm)

mm_scaled_data_set = pd.DataFrame(data_scaled_mm, columns=data.columns)
mm_scaled_data_set
```
