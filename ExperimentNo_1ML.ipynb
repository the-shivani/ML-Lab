```python
titanic.isnull().sum()
sns.heatmap(titanic.isnull(), yticklabels=False, cbar=False)
```

```python
# 1. Count plot: Gender count by Company
sns.countplot(data=titanic, x='Company', hue='Gender', palette='autumn')
plt.xticks(rotation=45)
plt.title('Gender Count by Company')
plt.show()
```

```python
# 2. Count plot: Employees by Country
sns.countplot(data=titanic, x='Country', palette='Set2')
plt.xticks(rotation=45)
plt.title('Employee Count by Country')
plt.show()
```

```python
# 3. Age Histogram
titanic['Age'].plot(kind='hist', bins=30, color='green')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.show()
```

```python

# 4. Salary Histogram
titanic['Salary'].plot(kind='hist', bins=30, color='orange')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.show()
```

```python
# 5. Scatter plot: Age vs Salary by Gender
sns.scatterplot(data=titanic, x='Age', y='Salary', hue='Gender')
plt.title('Age vs Salary by Gender')
plt.show()
```

```python
# 6. Scatter plot: Age vs Salary by Country
sns.scatterplot(data=titanic, x='Age', y='Salary', hue='Country')
plt.title('Age vs Salary by Country')
plt.show()
```

```python
# 7. Boxplot: Salary by Company
plt.figure(figsize=(10,5))
sns.boxplot(data=titanic)
```

```python
# Calculate correlation matrix for numeric columns only
correlation = titanic.corr(numeric_only=True)
print(correlation)

# View correlation specifically for Survived
print("\nCorrelation with Salary:")
print(correlation['Salary'])

# Visualize correlation matrix heatmap
plt.figure(figsize=(12, 7))
sns.heatmap(correlation, annot=True, cmap='crest')
plt.show()
```
